# IMDb Dataset — Exploratory Data Analysis

## Overview

This notebook explores the IMDb Top 1000 movies dataset to identify
basic patterns and characteristics related to movie revenue, IMDb ratings,
runtime, and release years.

The analysis focuses on descriptive statistics and visual exploration
of the dataset.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Load the Dataset

The dataset is loaded into a Pandas DataFrame for exploratory analysis.

```python
df = pd.read_csv("../data/imdb_top_1000.csv.xls")
```

## Movie Revenue Analysis

### Descriptive Statistics of Movie Revenue

#### Question

What are the minimum, maximum, and average gross revenues of the movies
in the dataset?

```python
df["Gross"] = (df["Gross"].str.replace(",","", regex=False).astype(float))

minimum_gross = df['Gross'].min()
maximum_gross = df['Gross'].max()
average_gross = np.round(df['Gross'].mean(), 3)

print(f"Minimum Gross: {minimum_gross}")
print(f"Maximum Gross: {maximum_gross}")
print(f"Average Gross: {average_gross}")
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df["Gross"], bins=20)

plt.title("Distribution of Movie Gross Revenue")
plt.xlabel("Gross Revenue")
plt.ylabel("Number of Movies")

plt.show()
```

#### Interpretation

The descriptive statistics and visualization provide an initial overview
of the distribution of movie gross revenue in the dataset...

## IMDb Rating Analysis

### Descriptive Statistics of IMDb Ratings

#### Question

What are the minimum, maximum, and average IMDb ratings of the movies
in the dataset?

```python
minimum_imdb = df['IMDB_Rating'].min()
maximum_imdb = df['IMDB_Rating'].max()
average_imdb = np.round(df['IMDB_Rating'].mean(), 3)
print(f"Minimum IMDB: {minimum_imdb}")
print(f"Maximum IMDB: {maximum_imdb}")
print(f"Average IMDB: {average_imdb}")
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df["IMDB_Rating"], bins=10)

plt.title("Distribution of IMDb Ratings")
plt.xlabel("IMDb Rating")
plt.ylabel("Number of Movies")

plt.show()
```

## Movie Runtime Analysis

### Average Movie Runtime

#### Question

What is the average runtime of the movies in the dataset?

```python
df["Runtime"] = (
    df["Runtime"]
    .str.replace(" min", "", regex=False)
    .astype(int)
)

average_runtime = df["Runtime"].mean()

print(f"Average Movie Runtime: {average_runtime:.2f} minutes")
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df["Runtime"], bins=15)

plt.title("Distribution of Movie Runtime")
plt.xlabel("Runtime (minutes)")
plt.ylabel("Number of Movies")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df["Runtime"], bins=15)

plt.title("Distribution of Movie Runtime")
plt.xlabel("Runtime (minutes)")
plt.ylabel("Number of Movies")

plt.show()
```

## Movie Release Year Analysis

### Oldest and Newest Movies

#### Question

Which movies are the oldest and newest in the dataset based on their
release year?

```python
df["Released_Year"] = pd.to_numeric(
    df["Released_Year"],
    errors="coerce"
)

oldest_movie = df.loc[df["Released_Year"].idxmin(), "Series_Title"]
oldest_year = df["Released_Year"].min()

newest_movie = df.loc[df["Released_Year"].idxmax(), "Series_Title"]
newest_year = df["Released_Year"].max()

print(f"Oldest Movie: {oldest_movie} ({int(oldest_year)})")
print(f"Newest Movie: {newest_movie} ({int(newest_year)})")
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df["Released_Year"].dropna(), bins=20)

plt.title("Distribution of Movie Release Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Movies")

plt.show()
```

## Conclusion

This notebook provided an exploratory analysis of the IMDb Top 1000 movies
dataset. We examined movie gross revenue, IMDb ratings, runtime, and release
years using descriptive statistics and visualizations.

The analysis provided an initial understanding of the main characteristics
and distributions of these variables. The findings from this exploratory
analysis can be used as a basis for further analysis of relationships and
