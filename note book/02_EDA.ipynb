# IMDB dataset Analysis

```python
import numpy as np
import pandas as pd
```

## Load the Dataset

```python
df = pd.read_csv("../data/imdb_top_1000.csv.xls")
```

## maximum , minimum and average gross

```python
df["Gross"] = (df["Gross"].str.replace(",","", regex=False).astype(float))

print(f"Minimum Gross: {df['Gross'].min()}")
print(f"Maximum Gross: {df['Gross'].max()}")
print(f"Average Gross: {np.round(df['Gross'].mean(), 3)}")
```

## maximum , minimum and average IMDB

```python
print(f"Minimum IMDB: {df['IMDB_Rating'].min()}")
print(f"Maximum IMDB: {df['IMDB_Rating'].max()}")
print(f"Average IMDB: {np.round(df['IMDB_Rating'].mean(), 3)}")
```

## Average Movie Runtime

```python

print(f"Average Movie Time: {df["Runtime"].mean()}")
    
```

## Oldest and Newest Movies

```python
oldest_idx = df["Released_Year"].idxmin()
print(f"Oldest Movie Name: {df.loc[oldest_idx, 'Series_Title']}")
newest_idx = df["Released_Year"].idxmax()
print(f"Newest Movie Name: {df.loc[newest_idx, 'Series_Title']}")
```
