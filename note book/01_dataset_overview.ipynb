# IMDB Dataset Overview

```python
import pandas as pd
```

## Dataset Overview

This notebook provides an initial overview and data quality assessment
of the IMDb Top 1000 movies dataset.

The goal is to understand the structure, data types, missing values,
and potential data quality issues before performing exploratory analysis.

## Load the Dataset

The dataset is loaded into a Pandas DataFrame for further inspection
and analysis.

```python
df = pd.read_csv("../data/imdb_top_1000.csv.xls")
```

```python
df.head()
```

## Initial Dataset Inspection

### Dataset Dimensions

First, we examine the number of observations and features in the dataset.

```python
rows, columns = df.shape

print("Number of rows:", rows)
print("Number of columns:", columns)
```

### Data Types and Non-Null Values

The `info()` method provides an overview of the dataset structure,
including column names, data types, and the number of non-null values.

```python
df.info()
```

### Statistical Summary

The `describe()` method is used to obtain descriptive statistics
for numerical variables in the dataset.

```python
df.describe()
```

## Data Quality Assessment

### Missing Values

Missing values can affect the quality of data analysis. Therefore, we
first identify the columns containing missing values and measure the
extent of missing data.

```python
missing_data = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_data = missing_data[missing_data["Missing Values"] > 0]

missing_data.sort_values(
    by="Missing Percentage",
    ascending=False
)
```

### Duplicate Rows

Duplicate rows can lead to biased results during analysis. Therefore,
we check whether the dataset contains any completely duplicated records.

```python
duplicated_rows = df.duplicated().sum()

print(f"Duplicated row count: {duplicated_rows}")
```

No completely duplicated rows were found in the dataset.
